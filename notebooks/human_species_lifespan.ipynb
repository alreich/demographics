## Answer

Yes — the notebook's total-ever-born estimate is exactly the kind of number that feeds into this, but it only answers one narrow piece of a much larger and genuinely contested question. Let me lay out how the pieces fit together.

**The core idea: this is a Doomsday Argument question**

The assumption you're describing — "we don't occupy a privileged position in the sequence of all humans" — is precisely the self-sampling assumption behind the Carter-Leslie Doomsday Argument, and its close cousin, J. Richard Gott's "delta-t" argument (which he applied to the Berlin Wall, Broadway shows, and the human species itself in a 1993 *Nature* paper). Both use a version of Copernican reasoning: if you're a randomly selected observer from the total population of humans who will ever exist, your birth rank (or your position in elapsed time) lets you place a probabilistic bound on how many more humans are coming (or how much longer the run continues).

There are two variants, and they use different inputs:

- **Carter-Leslie (rank-based):** Uses your position in the *cumulative count of births*, not time. This is exactly where the notebook's total (approximately 79-82 billion born so far, out of some unknown eventual total N) plugs in directly. If you assume your birth rank is a uniform random draw from 1 to N, you can derive a posterior on N (with a suitably chosen prior — often a Jeffreys or log-uniform prior over N, since the answer is sensitive to that choice).
- **Gott's delta-t (time-based):** Uses how *long* the species/phenomenon has already existed, not how many members it's produced. Applied to Homo sapiens, this would use "approximately 200,000-300,000 years since speciation" rather than "approximately 80 billion births," and gives a different kind of forecast — a future duration prediction rather than a future population size prediction.

Both are legitimate readings of your question, and they don't automatically agree with each other, because population size hasn't been constant over that history (we've had wildly different birth rates and population levels across eras) — so "you're randomly placed in the sequence of births" and "you're randomly placed in the timeline" are different assumptions with different implications. The notebook's model is built for the first one.

**What you get from the notebook, and what's still missing**

The notebook gives you N_past (total ever born so far) with a full posterior distribution rather than a point estimate — which is actually a nice fit for this kind of Bayesian doomsday calculation, since you could propagate that uncertainty into the downstream one instead of using a single number. But to go from "N_past posterior" to "time remaining before extinction or speciation," you need several more pieces that are outside anything a births-accounting model can give you:

1. **A future birth-rate/population trajectory.** The Carter-Leslie argument gives you a posterior on total N (past + future), which tells you *how many more births are coming*, not how much *time* that represents. Converting remaining-births into remaining-years requires assuming something about future fertility and population size — and this is arguably more uncertain than anything in the historical model, since it depends on demographic transition trends (most projections show global fertility converging toward or below replacement), technology, possible off-world expansion, etc. This is a genuinely different and harder modeling problem than reconstructing the past.

2. **A definition and model of "speciation."** This is a population-genetics question, not a demographic one. You'd need: effective population size (Ne), generation time, mutation rate, and — critically — the degree of reproductive isolation between subpopulations. Real allopatric speciation requires isolated breeding populations diverging over time; a single, huge, highly interconnected global population with constant gene flow (as we have now) is actually *resistant* to natural speciation, not prone to it. So "how long until we speciate naturally" may functionally be "not on any timescale shorter than a large isolating event (space colonization, catastrophic population fragmentation)," while "how long until we speciate via deliberate genetic engineering" is an entirely different, non-evolutionary question with no real historical base rate at all.

3. **A background extinction-rate model.** Absent the sudden catastrophes you're explicitly excluding, "time to extinction" is normally estimated from the fossil record using a *reference class* of comparable species — typically other mammals, or more narrowly other hominins — and looking at their observed durations from origination to extinction. This is genuinely a macroevolutionary/paleobiological question (Van Valen's "Law of Extinction," and later work by Foote, Raup, and others on taxon-duration distributions), and it runs into the classic **reference-class problem**: mammalian species durations average somewhere in the range of a very roughly 1-2 million years, but vary by orders of magnitude depending on body size, geographic range, and generation time — and hominin species specifically seem to have had shorter typical durations than that broader mammalian average. Which reference class you pick changes the answer enormously, and there's no principled way to choose one that isn't itself a modeling assumption.

**On extreme value analysis specifically**

This is a good instinct, but it's worth being precise about what EVA does versus what this question needs, because they're related but not the same tool.

Einmahl et al.'s approach works because they had a *large sample of repeated, roughly i.i.d. realizations* — many individual human lifespans — and they used extreme value theory to characterize the tail of that distribution and estimate its endpoint (the theoretical maximum age a human lifespan distribution asymptotes toward). That's a classic EVT setup: many draws from one distribution, asking about the extreme tail.

"How much longer will Homo sapiens persist" is structurally different: we have exactly **one** ongoing realization (this species, this lineage), not many repeated trials. You can't directly apply EVT to a sample size of one. What you *can* do is borrow the EVT toolkit at one level removed, in either of two ways:

- **Cross-species reference class as the "sample."** Treat the durations of many past species (all mammals, or all hominins, or some other taxon) as your i.i.d.-ish sample, and use EVT/order-statistics methods to characterize the tail of the *species-duration distribution* — effectively asking "given the distribution of how long species like us tend to last, what's the plausible upper bound, and where does 'time already elapsed' put us within that distribution?" This is a legitimate move, and it's conceptually close to what paleobiologists already do with survival/hazard analysis on fossil duration data — though EVT proper (fitting a GEV or GPD to block maxima or threshold exceedances) would be an unusual and stronger-than-necessary tool for that, since what you generally want is the *whole* survival function (hazard rate over time), not just the extreme tail. Standard survival analysis (Kaplan-Meier-style, or a parametric hazard model like Weibull) is the more natural fit for "time-to-event" questions; EVT is specifically about tail/maximum behavior.

- **Use EVT's output as an input, not as the whole method.** You could use an EVT-style analysis of the reference-class duration data to pin down a plausible ceiling on species duration, and then feed that as a prior or constraint into the Doomsday/Gott-style Bayesian update, which handles the "where are we positioned" part separately.

So: EVA isn't the wrong tool, but it answers a different sub-question (what's the tail behavior of a *distribution of durations across many species*) than the Doomsday argument does (given your specific rank/position, what does that imply about *this one* lineage). A complete treatment would likely combine both — an outside-view macroevolutionary/EVT-informed base rate, and an inside-view Bayesian self-sampling update using the notebook's population numbers — and then have an honest discussion of how much weight to put on each, given that both rest on genuinely contested philosophical assumptions (the self-sampling assumption has well-known critiques, e.g. Bostrom's "presumptuous philosopher" problem, and reference-class choice is unavoidably somewhat arbitrary).

That's the landscape as I see it — happy to go deeper on any one branch (the demographic-transition literature for future birth rates, the population-genetics side, or the paleobiological duration data) before we decide whether/how to build any of this out.

## Answer

Good — there's a substantial literature here, split cleanly into three tiers: the underlying data sources, methods developed specifically to correct for how unreliable that raw data is, and a separate, much thinner body of evidence specifically about hominins. Let me go through each.

## 1. Where the data comes from

**Large multi-taxon fossil-occurrence databases.** These are the backbone of essentially all quantitative macroevolutionary work:

- **Sepkoski's compendia** — the classic resource. Jack Sepkoski compiled first- and last-occurrence data for thousands of marine animal *families* (1982, 2nd ed. 1992) and later *genera* (2002), covering the whole Phanerozoic (~540 million years). This is what most of the historically famous results (extinction periodicities, background extinction rates, mass-extinction magnitudes) were built on.
- **The Paleobiology Database (PBDB, paleobiodb.org)** — the modern, continuously-updated successor. It's occurrence-based rather than range-based (it records "this taxon was found at this place in this time interval" rather than just first/last appearance), open and queryable, and now holds well over a million occurrences across 300,000+ fossil taxa. It's explicitly designed to let independent researchers run their own analyses rather than trusting a single curated summary.
- **Taxon-specific databases** — e.g., the **NOW Database (New and Old Worlds Database of Fossil Mammals)**, used for mammal-specific duration studies. These matter because marine invertebrates (Sepkoski's main strength) and terrestrial mammals have very different preservation dynamics, so a mammal-specific question needs mammal-specific data.

**Studies that directly estimate species/taxon durations from that data:**

- **Van Valen (1973)** — the foundational paper. He built survivorship curves for over 25,000 taxa across wildly different groups and found most decline roughly exponentially, implying a roughly *constant* extinction probability per unit time regardless of how long a taxon has already existed (his "Law of Extinction," which motivated the Red Queen hypothesis). This is directly relevant to your question: it's evidence *against* the idea that species accumulate "extinction risk" with age the way individual organisms do — a species that's persisted a long time isn't more (or less) likely to be near its end, at least by this model.
- **Prothero (2014)**, "Species longevity in North American fossil mammals," *Integrative Zoology* — probably the most directly citable modern figure for "typical mammal species duration": ~1–2 million years on average across most groups, with a widely-quoted Cenozoic-wide mean around 3.2 million years reported in a related 2013 GSA study, and enormous variance (some lineages exceed 20 million years; most are far shorter).

## 2. How reliable is this, really

This is the more important part of your question, and the honest answer is: reliable enough to support broad-strokes generalizations (orders of magnitude, rough averages), but not reliable enough to support a precise number for any one species — and hominins specifically are close to a worst case for several of these problems at once.

**Structural biases baked into any fossil-derived duration:**

- **The Signor-Lipps effect.** A species' *true* first and last appearance almost always precede/postdate its earliest and latest known *fossils* — you're sampling a fraction of the actual population through time, so the observed range is necessarily an underestimate, and rare or poorly-preserved taxa are hit hardest. This is one of the most well-established artifacts in paleontology.
- **"Pull of the Recent"** (Raup, 1979) — recent-era taxa look like they persisted longer than they really did (relative to how well we'd detect an equally long-lived ancient taxon), simply because sampling density and taxonomic scrutiny both increase toward the present, and living relatives extend the apparent range. This cuts directly against comparing "how long has *H. sapiens* lasted so far" to how long ancient taxa lasted — the measurement isn't on equal footing across time.
- **Taphonomic and habitat bias.** Preservation potential varies hugely by body size, skeletal composition, and environment. Hard-shelled marine invertebrates in depositional basins preserve beautifully; small-bodied, soft-tissue, tropical-forest-dwelling terrestrial animals — which describes most hominins — preserve terribly. The mammal fossil record is already much sparser than the marine invertebrate record it's usually benchmarked against, and primates within that record are sparser still.
- **Sampling heterogeneity over time.** Continental area exposed for collection, research effort, and rock outcrop availability all vary by era and region in ways that aren't uniform, which distorts raw counts if untreated.

**Statistical methods exist specifically to correct for this**, which is a genuinely useful development worth knowing about if you want to eventually use this data rigorously rather than just quoting a rule-of-thumb number:

- **Foote & Raup (1996)**, "Fossil preservation and the stratigraphic ranges of taxa," *Paleobiology* — models the *true* range as an unknown parameter and derives statistical confidence intervals on it from the gaps observed *within* a taxon's known fossil occurrences (more/denser internal occurrences → tighter, more trustworthy bounds on when it really originated/went extinct).
- **Marshall (1990, 1994)** and **Strauss & Sadler (1989)** developed related confidence-interval methods for stratigraphic ranges, treating the recovery-horizon problem as an order-statistics estimation problem.
- **Alroy and others' sampling-standardization methods** (used in PBDB's own quality-control pipeline) try to correct diversity/duration estimates for uneven sampling effort across time and place, precisely to counteract Pull-of-the-Recent-type effects.

The catch: **all of these correction methods need many fossil occurrences within the taxon's range to work well.** They shine for well-sampled marine invertebrate genera with hundreds of known localities. They're nearly useless for a taxon known from a literal handful of specimens — which is exactly the hominin situation.

**Why hominins specifically are a harder case than the "mammals last ~1-2 million years" statistic suggests:**

- **Sample size.** There are perhaps 10-20 recognized hominin species total, depending on classification (and see below) — nowhere near the "law of large numbers" comfort of Sepkoski's thousands of marine genera. Any duration statistic built from this set has enormous statistical uncertainty, and a couple of long-lived outliers (*Homo erectus*, at roughly 1.9-2 million years — by far the longest-lived member of our genus) dominate the average.
- **Taxonomic instability ("lumpers vs. splitters").** Whether *Homo ergaster*, *Homo heidelbergensis*, *Homo antecessor*, or *Homo rhodesiensis* count as separate species or as regional/temporal variants of one lineage is genuinely unsettled among paleoanthropologists, and different published classifications give different total counts and different individual durations for the very same fossils. This isn't a sampling problem, it's a conceptual one: the biological species concept (reproductive isolation) is essentially unobservable in fossils, so paleontologists rely on morphological criteria that different researchers weight differently.
- **Individual species dates keep moving.** *Homo erectus*'s end date has itself shifted materially with better dating methods — the Ngandong, Indonesia population was for a long time thought to have persisted to ~35,000-50,000 years ago, but a 2019 uranium-series/ESR redating study pushed that back to ~117,000-108,000 years ago. If the best-known, most scrutinized hominin species' duration estimate moved by tens of thousands of years within the last decade, that's a useful calibration for how much uncertainty band to put around any of these numbers.

**Bottom line for how you'd actually use this:** the "mammals last ~1-2 million years, sometimes up to 10+" statistic is a reasonably solid outside-view anchor precisely because it's built from large-n, well-corrected data. A hominin-specific version of that statistic is much shakier — small n, contested taxon boundaries, worse preservation, and a record still being actively revised. That doesn't make it useless (it's still better than a guess), but it means any duration estimate should carry a visibly wide uncertainty band, and probably should be framed as "consistent with the broader mammalian pattern" rather than "precisely estimated from hominin-specific data."

## Answer

Genuinely mixed, and worth being precise about which parts of the "time remaining" question this work actually speaks to versus doesn't. Here's how I'd break it down.

## Pros

**It's real, large-sample, censoring-corrected data — not a quoted point estimate.** Going from "mammals last ~1-2 million years" (a number lifted from a review paper) to an actual empirical survival curve built from 12,469 species, with extant species properly treated as right-censored rather than dropped or miscounted, is a genuine improvement in rigor. It gives you a full distribution to propagate uncertainty through, rather than a single number to anchor on.

**The mathematical structure matches the question unusually well.** "Time remaining" is fundamentally a residual-life question — given a species has already survived to age T, what's the distribution of additional time before it exits (via extinction or, as I'll come back to, possibly via evolving into something else)? That's exactly what a survival/hazard function is built to answer. This fits the question's actual shape better than, say, the Doomsday argument's birth-rank framing does, which answers a related but structurally different question.

**It's a legitimate "outside view."** Grounding a forecast in a broad reference class of comparable historical cases, rather than reasoning from the inside about what feels special or different about the present case, is good forecasting practice (this is the core idea behind Kahneman's "outside view" and Tetlock's superforecasting work). A species-duration base rate is a reasonable instance of that.

**Two priors that agree is a real (if limited) reliability signal.** The Kaplan-Meier and log-logistic fits landed on similar quantiles independently — one nonparametric, one a specific functional form. That's not nothing; it means the answer isn't hypersensitive to that particular modeling choice.

**It's reusable, extensible infrastructure.** The notebook, sampler functions, and hazard fits can be fed directly into a later model as an informative prior, rather than needing to be rebuilt.

## Cons

**Reference-class mismatch is the big one.** Homo sapiens is an outlier on nearly every axis that plausibly affects extinction or speciation risk relative to the other ~12,469 species in this dataset: global range instead of typically restricted range, population in the billions instead of orders of magnitude smaller, one massively interconnected breeding population instead of semi-isolated subpopulations, and an unprecedented degree of deliberate environmental modification (niche construction, technology-mediated buffering against many traditional extinction pressures). A distribution built mostly from small-bodied, geographically-restricted mammals may simply not transfer. This is a real, unresolved version of the classic reference-class problem: which set of "similar cases" is actually similar in the ways that matter?

**A body-size-stratified version would probably be better, and we haven't built it.** There's documented evidence (the Liow/Finarelli-type work we found earlier) that origination/extinction dynamics differ systematically between large- and small-bodied mammals. Humans are large-bodied; the aggregate "all Mammalia" distribution is dominated by small-bodied taxa (rodents, etc.) simply because they're more numerous. Filtering to large mammals, or primates specifically, would be a natural refinement — at the cost of a much smaller, noisier sample.

**There's a real tension with your own framing of the question.** You asked about time remaining "assuming nothing more sudden and tragic happens, like world war or catastrophic climate change" — i.e., you want a background/gradual rate, excluding catastrophic causes. But most real extinctions in the fossil record were *not* gradual senescence-like processes — they were caused by exactly this class of event at the species level: climate shifts, competitive replacement, habitat loss, disease, localized catastrophe. Nature doesn't really have a clean "background extinction" channel that's separate from "sudden events" the way your question wants to carve it — that's a human analytical distinction we're imposing on the data, and it's not obvious the historical hazard rate we fit is actually estimating the same thing you're asking about.

**Timescale mismatch.** The resolution-floor problem from the notebook means this dataset fundamentally cannot resolve anything below roughly the width of a geological stage — typically hundreds of thousands of years, sometimes millions. Any human-relevant "time remaining" question on the scale of centuries to a few millennia is simply below what this data can speak to. At best, this approach can say something like "on a several-hundred-thousand-to-few-million-year timescale, is the human lineage's duration-so-far typical or unusual for a mammal" — not anything about the near or medium term.

**Non-stationarity across a huge span of Earth history.** The observed durations span up to ~119 million years, across wildly different climatic regimes, continental configurations, competitor communities, and background extinction pulses (including mass extinctions). Using an aggregate hazard fit implicitly assumes some kind of stationarity — that "the rate at which species like this exit" is a meaningful constant to extrapolate — when the modern, human-dominated biosphere may not resemble any of the conditions that generated the training data.

**Sampling bias that specifically distorts the recent end of the curve.** We already flagged "Pull of the Recent" in the literature review — recent-era taxa look artificially long-lived because of denser sampling and living descendants extending the apparent range. Since Homo sapiens' entire existence sits at the very recent end of this timeline, any inference that leans on how the "recent" portion of the curve behaves is exactly where this bias is worst.

**The chronospecies/pseudo-extinction problem cuts both ways.** The fossil record often can't distinguish a lineage going truly extinct from a lineage gradually evolving into a named descendant species (anagenesis) — paleontologists draw that boundary somewhat subjectively. In one sense this is convenient for you, since your question already lumps "extinction" and "evolving into a different species" together as the same kind of exit event, and the fossil data's ambiguity matches that framing. But it also means we can't currently decompose the fitted hazard into "probability of a true dead-end" versus "probability of an evolutionary transition," which you might eventually want to treat differently (they have very different implications).

**Two epistemically different tools, not obviously combinable.** This survival-curve approach is an outside-view, frequentist-style base-rate argument. The Doomsday/Gott argument from our first discussion is an inside-view, anthropic self-location argument. They can give different, even contradictory answers, and there's genuine, unresolved philosophical disagreement (Bostrom's self-sampling vs. self-indication assumption debates) about whether and how to combine anthropic reasoning with ordinary population base rates. Building both doesn't automatically tell you how to weigh them against each other.

**Agreement between the two priors is narrower evidence than it looks.** The KM and log-logistic fits agreeing is reassuring *within* one fixed set of choices (all-Mammalia reference class, `dur_max` as the duration measure). It says nothing about whether a different reference class or duration definition would give a similar answer — and we have good reason to think those choices matter a lot (see the body-size point above, and the `dur_max` vs. `dur_min` vs. `dur_mid` divergence we found earlier).

## Where that leaves it

None of this makes the work useless — it makes it one input rather than an answer. The honest framing is: this gives a defensible order-of-magnitude outside-view anchor (something like "durations in the low-single-digit-to-tens-of-millions-of-years range are typical for lineages broadly like this one"), useful as a sanity check and as a prior to be updated by other evidence — not a decision-grade forecast specific to humans' actual trajectory. Used well, it'd sit alongside the Doomsday-argument calculation and whatever we eventually do on the population-genetics/speciation-feasibility side, with the disagreements between them treated as informative rather than papered over.
