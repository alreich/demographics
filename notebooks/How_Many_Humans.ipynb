This is a great problem to think through before touching the numbers — the estimation is straightforward arithmetic once you fix your assumptions, but the *honest* uncertainty analysis is genuinely hard, largely because the sources of error are almost entirely systematic/epistemic rather than sampling error. Here's the landscape.

## What the estimate actually depends on

The canonical formulation (going back to Fucks and to Keyfitz) is that the cumulative number of humans ever born depends on essentially three things per historical period, integrated over time:

1. **When to start the clock** — anatomically modern *Homo sapiens* vs. earlier hominins vs. "behaviorally modern" humans. This single choice moves the answer by billions: PRB's own figure jumped from ~108 billion to ~117 billion largely because they pushed the starting date back from 50,000 BCE to roughly 190,000 BCE as the genetic/paleoanthropological consensus shifted.
2. **Population size at each point in time** — reconstructed from archaeological carrying-capacity estimates for prehistory, then increasingly from tax/census/parish records, and finally from UN World Population Prospects for the modern era. The classic source demographers lean on for the pre-modern trajectory is McEvedy & Jones's *Atlas of World Population History*; Michael Kremer's 1993 QJE paper "Population Growth and Technological Change: One Million B.C. to 1990" is another widely-cited compilation covering the very long run.
3. **Crude birth rate per period** — not the death rate. This is the counterintuitive part: the estimate does not depend on the number of deaths during any period of time — you only need births per 1,000 population per era, which in turn requires assumptions about life expectancy (to back out plausible birth rates from population growth rates, since in a low-growth regime births ≈ deaths ≈ some function of life expectancy). PRB's method effectively starts with just two people and grows a birth-rate-driven population forward.

The compounding difficulty is that population size didn't grow smoothly — it fluctuated wildly in response to famines and climate change, and life expectancy in most of human history was extremely low (PRB notes it may have averaged only about 10 years for much of human history, driven by high child mortality, requiring correspondingly implausible-sounding birth rates around 80 live births per 1,000 people per year to sustain any population growth at all). Every one of these inputs is a modeled reconstruction, not a measurement.

## Why this resists a clean frequentist CI

There's no repeated-sampling frame here — you're not drawing a sample from a population of "possible Earths." So a textbook frequentist confidence interval doesn't have an obvious referent. What people sometimes do instead is a **sensitivity analysis** (vary each assumption across its plausible range, report the resulting spread) — this is what Haub/PRB gesture at informally but never formalize with actual bounds.

A **Bayesian framing fits much better**: put priors on the uncertain quantities (start date, per-era population sizes, per-era birth rates — probably log-normal or Beta-ish priors informed by the ranges different sources give), propagate them through the same accounting identity via Monte Carlo, and report a posterior credible interval on cumulative births. This also lets you make the "3 factors" structure explicit as a hierarchical model rather than a point-estimate chain multiplication, and naturally handles the fact that uncertainty is dominated by a handful of periods (the deep Paleolithic, and the population at first agriculture ~8000 BCE) rather than spread evenly across all of history.

So: widely cited, but it's a single research lineage's back-of-envelope calculation, never presented with formal error bars, and it's an order-of-magnitude/"rough idea" exercise by its authors' own description, not a rigorously validated statistical estimate. Some critics find the figure far too small given uncertainty about early population sizes, which underscores that there's real disagreement even among people using similar methods.

## Why not just pick one wholesale

The two families aren't really competing *general* choices — they encode different epistemic claims, and those claims are true for some of your parameters and false for others.

**Beta-ish (hard bounds) is right when a real ceiling exists and you basically know where it is.** Crude birth rate is the clearest case: sustained CBR has a genuine biological ceiling around 55–60 per 1,000 (the highest sustained national rates ever recorded, e.g. Niger-era outliers, top out well below the ~95 upper bound log-normal will occasionally sample). There's no serious scientific argument for CBR = 90 for centuries — that's not "rare but possible," it's essentially impossible. Population at well-attested anchors (1850, 1900, 1950…) is similar: we have converging independent lines of evidence, so a hard, fairly tight band is honest, not overconfident.

**Log-normal (open tail) is right when the "bound" itself is really just current best belief, and the field has been wrong about it before.** The starting date is the textbook case — the PRB estimate's own headline number moved from ~108B to ~117B specifically because the consensus start date moved from 50,000 BCE to ~190–200,000 BCE. That's not noise within a known range; that's the entire frame shifting because of a fossil find. Forcing a hard wall at 200,000 BCE would repeat exactly the mistake that already happened once. Deep-prehistory population size (founding population, 8000 BCE) arguably belongs here too — archaeological carrying-capacity estimates for the Paleolithic are reconstructions from very sparse evidence, not measurements, so there's real "we might be off by more than we think" risk, not just sampling noise within stated bounds.

**So the honest per-parameter answer is:**
- Population and birth rate for the last ~2,000–4,000 years (where we have written records, censuses, tax rolls) → **Beta-ish**, tight bounds, high confidence the truth is inside them.
- Founding population, Paleolithic/early-Neolithic population and birth rate, and the starting date → **log-normal**, because the tail risk is epistemic (paradigm-shift risk), not just parameter noise.

This also directly addresses the mechanism you saw empirically: almost all of the log-normal interval's extra width (69–106B vs. 75–96B) is coming from a handful of ancient, poorly-constrained parameters compounding, not from the well-attested modern ones. A hybrid lets you keep log-normal's honesty exactly where it's earned, without paying for it on parameters that don't need it.

## A third option worth adding: truncated log-normal

Rather than a binary choice, consider a middle distribution for parameters where you want log-normal's realistic multiplicative shape *but* you do have one hard piece of information you're not willing to violate (e.g., "population in 1 CE definitely didn't exceed 500 million — that's archaeologically indefensible, full stop, but I don't want to commit to a symmetric Beta shape inside the bound either"). A **log-normal truncated at a stated ceiling** (PyMC: `pm.Truncated(pm.LogNormal.dist(mu, sigma), upper=cap)`) gives you the calibration convenience of log-normal (same `mean, low, high → mu, sigma` helper you already have) while removing just the implausible extreme tail, rather than replacing the whole shape with Beta's more artificial symmetric-ish mound. It's a genuinely useful third option specifically for the awkward middle-ground parameters — I'd reach for it on things like 8000 BCE and 3000 BCE population, where you believe log-normal's skew is realistic but the pure unbounded tail occasionally produces numbers larger than the land could plausibly have supported.

## Bottom line

- If you must ship one uniform choice today: **Beta-ish**, because most of your 25 parameters really do have defensible bounds, and it produces a tighter, more publishable interval without hand-waving away implausible extremes.
- Better: **hybrid** — Beta-ish for the well-attested recent parameters, log-normal (or truncated log-normal) for the handful of deep-history/paradigm-risk parameters.
- Keeping both full parallel models running indefinitely isn't worth the overhead once you understand the mechanism, as we now do — its main remaining value is as a one-time robustness check you can point to in a writeup ("the conclusion doesn't hinge on this modeling choice"), not something to maintain going forward.
